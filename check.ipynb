```python
from qiskit import QuantumCircuit

# Create a Quantum Circuit with 1 qubit and 1 classical bit
qc = QuantumCircuit(1, 1)

# Apply Hadamard gate to put the qubit in a 50/50 superposition
qc.h(0)

# Measure the qubit and store the result in the classical bit
qc.measure(0, 0)

# Print a text-based circuit visualization
print("Circuit built successfully:")
print(qc.draw('text'))
```

```python
print("Fetching real hardware backends and checking their queues...\n")

# Get all real, operational hardware systems
available_backends = service.backends(simulator=False, operational=True)

# Loop through and print the status and queue count for each
print(f"{'Backend Name':<20} | {'Status':<12} | {'Pending Jobs (Queue)':<20}")
print("-" * 60)

for b in available_backends:
    status = b.status()
    # status.status_msg tells you if it's active/operational
    # status.pending_jobs tells you how many jobs are in line
    print(f"{b.name:<20} | {status.status_msg:<12} | {status.pending_jobs:<20}")
```
